# Here is how you use the bay package

Nomenclature:

- RUN is a full execution of a recipe. `analysis` will have components meant to do analysis on a full run (like comparing mcmc stuff between chains) or a single chain (for doing the actual bayesian analysis).
- CHAIN is a single MCMC (`emcee`) execution of a single config in the recipe (we do like 1-30 chains per run)
- TRANSIT is a single set of parameters with their associated light curve and transit. U can use an instance of `TransitModel` (just check which parameters it takes) and get all the parameters to compute a batman light curve.

```python
import bay
```

literally just run this and it gives you the `.npy` file that u can do analysis on:

```python
bay.run(recipe_path="recipes/prior_sensitivity.yaml", 
        data_path="data/apjladf06ft3_mrt.txt",
        output_path="output")
```
