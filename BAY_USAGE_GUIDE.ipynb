# Here is how you use the bay package

Nomenclature:

- RUN is a full execution of a recipe. `analysis` will have components meant to do analysis on a full run (like comparing mcmc stuff between chains) or a single chain (for doing the actual bayesian analysis).
- CHAIN is a single MCMC (`emcee`) execution of a single config in the recipe (we do like 1-30 chains per run)
- TRANSIT is a single set of parameters with their associated light curve and transit. U can use an instance of `TransitModel` (just check which parameters it takes) and get all the parameters to compute a batman light curve.

```python
import bay
```

literally just run this and it gives you the `.npy` file that u can do analysis on:

```python
bay.run(recipe_path="recipes/prior_sensitivity.yaml", 
        data_path="data/apjladf06ft3_mrt.txt",
        output_path="output")
```

once you have the files you can check what `analysis.py` is packing. below are some examples:

```python
# the intended behavior TODO

run_analysis = bay.analysis.run(run_path="output/2026-11-18_1321_prior-sensitivity")
chain_analysis = bay.analysis.chain(chain_path="output/2026-11-18_1321_prior-sensitivity/2026-11-18_1321_prior-sensitivity")
transit_analysis = bay.analysis.transit(chain_path="output/2026-11-18_1321_prior-sensitivity/2026-11-18_1321_prior-sensitivity", transit_idx=50) # or best transit or something

run_analysis.compare_params(["tau, eccentricity", "period"])
chain_analysis.plot_corner()
chain_analysis.get_params() # for the statistical summary of the chain etc
transit_analysis.plot_orbit()
transit_analysis.plot_transit()
transit_analysis.get_transit_params()
```
